```python
# Now we're onto the meat and potatoes of coding with Conditionals and Loops :)
# If/else is relatively self-explanatory.

DNA_seq = 'ATGTAGGATATAGC'

codon1 = DNA_seq[0:3]

if (codon1 == 'ATG'):
    print('Codon', codon1, 'is a start codon.')
else:
    print('Codon', codon1, 'is not a start codon.')
```

```python
# Elif is if you want to mutual exclusivity with multiple if statements,

DNA_segment = 'ATGACATGACCAATC'
codon1 = DNA_segment[-3:]

if (codon1 == 'ATG'):
    print('Codon', codon1, 'is a start codon.')
elif ((codon1 == 'TAA') or
      (codon1 == 'TAG') or
      (codon1 == 'TGA')):
    print ('Codon', codon1, 'is a stop codon.')
else:
    print('Codon', codon1, 'is neither a start nor a stop codon.')
```

```python
# On to our first loop, the while loop continues to loop while an expression remains true

DNA_seq = 'CGGACACACAAAAAGAATGAAGGATTTTGAATCTTTATTGTGTGCGAGTAACTACGAGGAAGATTAAAGA'
print('DNA sequence:', DNA_seq)
 
bp = 'T'
print('Base pair:', bp)

print('# of Ts in DNA_seq via .count(): ', DNA_seq.count(bp))

count = 0
index = 0

while index < len(DNA_seq):
    if bp == DNA_seq[index]:
        count += 1
    index += 1


print ('# of Ts in DNA_seq via while loop: ', count)
```

```python
# The for loop continues for each element in a sequence

bases = ['T', 'C', 'A', 'G']

for base in bases:
    print(base, end='')

# Let's say we want to print all combinations of codons that start with T, we could use a for loop to generate all iterations of the first two letters
# Then after generating the iterations of the first two letters, follow up with all iterations of the third nucleotide
print ('\n'*2 + 'All codons that start with T are as follows: '+'\n')

for ntd2 in bases:
    print ('Codons starting with T'+ntd2+':')
    for ntd3 in bases:
        print('T'+ntd2+ntd3)
```

```python
# The range function allows generation of a list of a integers for you to access you can use it to solve this problem:
# Assume the population of Florida sandhill cranes grows by  1.94% annually. If we start with a population of 425 birds, 
# how large will the population be after 30 years, calculate using a for loop.
```

```python
population = 425
growth_rate = 0.0194
years = range(30)

for year in years:
    population = population+(population*growth_rate)

print(int(population))
```

```python
# lets say you wanted to model the growth throughout years

population = [425]
growth_rate = 0.0194
years = range(30)

for year in years:
    next_gen_pop = population[year]+population[year]*growth_rate
    population.append(next_gen_pop)

# In the case we want to use this population nums for future uses, we can store it as a list and append to it as a variable
# Otherwise you can use a temporary variable which saves on memory

for year in years:
    print('The population of Florida Sandhill cranes at ' + str(year) + ' years is ' + str(round(population[year])))
    
```

```python
# Testing what you know, use a dictionary to count the number of times a codon appears in a segment
DNA_seq = 'gggtgcgacgattcattgttttcggacaagtggataggcaaccactaccggtggattgtc'
print('Sequence:', DNA_seq)

codons = [] # split the string into groups of 3 and store into this list
for i in range(0, len(DNA_seq), 3): # we can generate a list of index points to mark the positions of the codons then slice at those points
    codons.append(DNA_seq[i:i+3])

codon_tracker = {}
for codon in codons:
    if codon not in codon_tracker:
        codon_tracker[codon] = 1
    else:
        codon_tracker[codon] += 1

for codon, count in codon_tracker.items():
    print(codon, ':', count)
```
