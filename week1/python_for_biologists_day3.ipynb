```python
# The .find() function searches for substrings in a string, you can search in a range or the entire string
# .rfind() allows you to search starting from the end of the string without reversing the original string
# count() will never return an error
# index() is used when you expect something to be there and want to know if its not there

# i just learned that convention is constant variables should be upper snake case sooo
CHIMP_DLOOP = 'GTACCACCTAAGTACTGGCTCATTCATTACAACCGGTATGTACTTCGTACATTACTGCCAGTCACCATGA'
codon = 'CAT'

codon_check = codon in CHIMP_DLOOP
print (codon_check)

codon_count = CHIMP_DLOOP.count(codon)
print(codon_count)

codon_find = CHIMP_DLOOP.find(codon)
print(codon_find)

# testing fail find instance
codon_findfail = CHIMP_DLOOP.find(codon, 0, 0)
print(codon_findfail)

codon_rfind = CHIMP_DLOOP.rfind(codon)
print(codon_rfind)
```

```python
# review of slicing is that it takes indices have a string start from [0] whereas negative indices start from [-1]
# the slice function takes a start, end, and step start is inclusive, end is exclusive, step can indicate direction

human = 'TTCTTTCATGGGGAAGCAGATTTGGGTACCACCCAAGTATTGACTTACCCATCAACAACCGCTATGTATT'
print('Human D-loop:', human)
 
mycodon = 'CAT'
 
# Find the lowest index of mycodon in human
index_mycodon = human.find(mycodon)
print('First', mycodon, 'index:', index_mycodon)
 
# Extract the first codon after mycodon
first_codon = human[index_mycodon + 3: index_mycodon + 6]
print('First codon after', mycodon, ':', first_codon)
 
# Extract the second codon after mycodon
second_codon = human[index_mycodon + 6: index_mycodon + 9]
print('Second codon after', mycodon, ':', second_codon)
 
# Negative starting point counts from the end of the string.
next_to_last_codon = human[-6:-3]
print('Next to last codon:', next_to_last_codon)
 
# Omitted second entry in slicing indicates to the end of string
last_codon = human[-3:]
print('Last codon:', last_codon)
```

```python
# The translate function relies on .maketrans() to produce a dictionary specifically for this translation purpose
# To translate, you first make the dictionary of how you want to translate e.g. ATCG, TAGC means A->T, T->A, C->G, G->C
# Then you use the .translate() function on the string you would like to translate and input the dictionary to be used in the brackets

ZIKA_DNA_SEQ = 'ATCGATTATCGAGTAGCTAGTCAT'
complement_trans = str.maketrans('ATCG', 'TAGC')
zika_cDNA = ZIKA_DNA_SEQ.translate(complement_trans)
print("Complementary sequence: " + zika_cDNA)

# Reversing a string is quite easy, it is just using a slice but reversing the step order so it outputs the string in rereverse

zika_rDNA_seq = ZIKA_DNA_SEQ[::-1]
print("Reverse sequence: " + zika_rDNA_seq)

# .translate() is quite quick for single base pair swapping + mapping, but is limited in phrase and length, in order to do larger batches of swaps
# we can use str.replace()

dengue_DNA = '601 catgtgtgac gccaccatga gttatgagtg'
dengue_onespace_del = dengue_DNA.replace(" ","", 1)
print (dengue_onespace_del)
dengue_allspace_del = dengue_DNA.replace(" ","")
print (dengue_allspace_del)
```
