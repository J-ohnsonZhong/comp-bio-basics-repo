```python
# 4. Reverse complement: Create a Python program that computes the reverse complement
# of the DNA sequence below and prints the sequence and its reverse complement: GACGTCGCCAGAGAggcataTAACGATAtgacacagagagagcaGAGACAAGT
```

```python
DNA_seq = "GACGTCGCCAGAGAggcataTAACGATAtgacacagagagagcaGAGACAAGT"
DNA_seq = DNA_seq.upper()
cDNA_seq = DNA_seq[::-1]

print("DNA sequence: " + DNA_seq)
print("Complement sequence: " + cDNA_seq)
```

```python
# 5. DNA to RNA: Write a Python program that prints out the corresponding RNA sequence of a given DNA sequence. 
# Make sure to handle sequences both in lowercase and uppercase (map t -> u, T -> U). 
```

```python
DNA_seq = input()
rnatrans = DNA_seq.maketrans('Tt', 'Uu')
DNA_seq = DNA_seq.translate(rnatrans)
print(DNA_seq)
```

```python
# 6. Counting base pairs: Write a Python code that prints out the percentages of the four nucleotides A, C, G, T in the following segment 
# of the breast cancer gene BRAC2: gggtgcgacg attcattgtt ttcggacaag tggataggca accactaccg gtggattgtc 
```

```python
BRAC2_seq = "gggtgcgacg attcattgtt ttcggacaag tggataggca accactaccg gtggattgtc"
num_a = BRAC2_seq.count('a')
num_t = BRAC2_seq.count('t')
num_g = BRAC2_seq.count('g')
num_c = BRAC2_seq.count('c')

total_ntd = num_a + num_t + num_g + num_c
print (f"Percentage of a: {num_a/total_ntd}", 
       f"Percentage of t: {num_t/total_ntd}",
       f"Percentage of g: {num_g/total_ntd}",
       f"Percentage of c: {num_c/total_ntd}",
       sep = '\n'
      )
```
